# *Python operators* :-

# *Arithmetic Operators:-*
* Used for mathematical calculations.
- (+)adds values.
- (-) subtracts the second value from the first.
- (*) multiplies values.
- (/) always gives a division result, usually a float.
- (%)gives the remainder.
- (**) calculates power.
- (//) removes the decimal part by floor division.

```python
a = 10

b = 3

print(a + b)    # 13
print(a - b)    # 7
print(a * b)    # 30
print(a / b)    # 3.333...
print(a % b)    # 1
print(a ** b)   # 1000
print(a // b)   # 3
```

```python
print(2**3) # 2 * 2 * 2
```

```python
print ( 2 ** 2 ** 3 ) 
```

```python
x = 5
y = 3 

print( x / y )
print( x // y )
print( x % y )

print( y / x )
print( y // x )
print( y % x )
```

# *Comparison Operators*:-
* Used to compare two values.
* The result is always True or False.
- (==) checks value equality.
- (!=) checks whether values are different.
- (>) means left value must be greater.
- (<) means left value must be smaller.
- (>=) means greater or equal.
- (<=) means smaller or equal.
- Do not confuse = with ==.

```python
print(a == b)   # False
print(a != b)   # True
print(a > b)    # True
print(a < b)    # False
print(a >= b)   # True
print(a <= b)   # False
```

```python
num1 = 5
num2 = 11

print(num1 == num2)   # Equal to
print(num1 != num2)   # Not equal to
print(num1 > num2)    # Greater than
print(num1 < num2)    # Less than
print(num1 >= num2)   # Greater than or equal to
print(num1 <= num2)   # Less than or equal to
```

```python
"tech" == "tech" # both are same or not 
```

# *Logical Operators*
* and , or , not , xor
* Rules for and :-

Both conditions must be True. 

True  and True  = True

True  and False = False

False and True  = False

False and False = False

* Rules for or:-

At least one condition must be True.

True  or True  = True
 
True  or False = True
 
False or True  = True

False or False = False


* Rule for not:-

It reverses the Boolean value.

not True  = False

not False = True

```python

a = 10
b = 5

condition1 = (a > b)
condition2 = (b < 20)

# Logical AND
result = condition1 and condition2

print(f"logical and result: {result}")
```

# *Identity Operators*
- (is) checks object identity or same objects.
- (==) checks value equality Different objects.
- For normal value comparison, use ==.
- is is commonly useful with None

```python
f = 5
g = 5

print("f is g:", f is g)
print("g is f:", g is f)
print("f is not  g:", f is not g)
print("g is not  f:", g is not f)

# This is a bat, cat
# She is tall girl.
# He is play boy.
# It is a animal.
```

# *What is slicing*:-
Slicing is a extract specific parts of code or sequence of code like String,List and Tuple.

Start: The beginning index (inclusive)

Stop: The ending index(exclusive).

Step: The number of items to skip (default is 1).

```python
text = "Use the in operator to check whether a word exists in a sentence."
print(text)
```

```python
text[0:19]
```

```python
text[0:10:5]
```

```python
a = [1,2,3,4,5,6,7,8,9]
print(a)
b = a[2:]
print(b)
c = a[:3]
print(c)
```
