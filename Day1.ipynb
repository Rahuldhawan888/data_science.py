# *Python Comment* :-

* In Python, comments are non-executable lines of text used to explain code and improve readability. The Python interpreter ignores them during execution.

* There are two primary ways to create comments:

* Single-line Comments
* Single-line comments start with the hash symbol (#). Anything following the hash symbol to the end of the line is considered a comment

```python
# This is my first comment :- 
print("Python bro you are so crazy !!!") 
print(2 + 2) # () => Paranthesis
print(6 - 2) # line by line
print(3 * 3) # line by line
print(3 / 3)  # line by line
print(4 % 3)  # line by line
print(2 ** 2)  # line by line
```

* Multiline Comments Python does not have a dedicated syntax for multiline comments like other languages (e.g., /* ... */ in C or Java).

*However, there are two common work arounds:*

* Multiple # symbols: The most straightforward way is to use a hash symbol at the beginning of each line.

*complex* – numbers with real and imaginary parts (e.g., 2+3j)
