```python
# 1. Student Names

# - A teacher stores the names of students in a class:
# Questions:
# - What type of data structure is students?
# - How do you access "Priya"?
# - How do you add "Rohan"?
# - How do you remove "Amit"?
# - How do you find the number of students?

students = ["Rahul", "Amit", "Priya", "Neha"]
print(type(students)) 
print(students[2])

students = set(students)
print(type(students)) 
students.add("rohan")
print(students)
students.remove("Amit")
print(len(students))
print(students)
 
```

```python
# Employee Record

# - An employee's basic information should not normally change:

# - employee = ("Rahul", 101, "Data Scientist")

## Questions:

# - What data structure is this?
# - How do you access the employee ID?
# - Why might a tuple be better than a list here?
# - Can you change "Rahul" to "Amit"? 

employee = ("Rahul", 101, "Data Scientist")
print(type(employee))
print(employee[1])
employee = list(employee)
employee[0] = "Amit"

print(employee)

```
