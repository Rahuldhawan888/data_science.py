# *Pyhton functions*
- Python functions are used for these important reasons:
1. Code Reusability — Write code once and use it many times.
2. Reduce Repetition — Avoid writing the same code again and again.
3. Easy Debugging — Find and fix errors more easily.
4. Modularity — Divide a large program into smaller parts.
5. Readability — Make code easier to understand and maintain.

```python
dic = {'id':[101,102,103,104],
      "Product":['A','B',"C",'B'],
       "Price":[1000,500,499,479],
       'Order Date':['2026-05-15','2026-05-15','2026-05-13','2026-05-12']}
print(dic)
```

```python
for i in dic:
    print(i)
```

```python
for i in dic:
    for j in dic[i]:
        print(j,end=" ") 
    print()
    
    
```

```python
for key,value in dic.items():
    print(key,value)
```

```python
# def keyword , function name , ()=> Parantheses

def my_fun(): # Defined the Function 
    # Prime Number ? 
    val = int(input()) 
    
    for i in range(2,val):
        if val % i==0:
            print(f"This number {val} is Not Prime Number")
            break
        else:
            print(f"This number {val} is Prime Number")
            break
```

```python
my_fun() # Calling a function :- 
```

```python
def my_fun(val): # Defined the Function  with parameter value 
    for i in range(2,val):
        if val % i==0:
            print(f"This number {val} is Not Prime Number")
            break
        else:
            print(f"This number {val} is Prime Number")
            break
```

```python
# Prime Number ? 
val = int(input()) 

res = my_fun(val)
```

```python
my_fun(7)
```

```python
def found_sales(sales,value): # defined the function:-
  if value in sales:
    return "Sales is Found"
  else:
    return "No Sales Found"
```

```python
sales = [23,65,87,54,87,23] # Number of Sales
val = float(input())
```

```python
res = found_sales(sales,val) # Calling Funciton 
print(res)
```
