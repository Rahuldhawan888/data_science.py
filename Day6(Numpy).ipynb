Numpy :- A one-dimensional labeled homogeneous array, size immutable.

```python
# ndarray vs list

a = np.array([12,56.2,78,56,98],dtype=float)
print(a)

a = np.array([12,56.2,78,56,98],dtype=str)
print(a)

a = np.array([12,56.2,78,56,98],dtype=int)
print(a)
```

```python
print(a.dtype) # Data types
```

```python
a = np.array([12,56.2,78,56,98],dtype='float16')
print(a)

a = np.array([12,56.2,78,56,98],dtype=str)  # String by- default 1Bypes
print(a)

a = np.array([12,56.2,78,56,98],dtype='int8')
```

```python
print(a.dtype)
```

# *Numpy dimensional*

- Dimensions determine how
- data is organized( eg.. )

0 -> D: Scaler -> np.array(5)

1 -> D: Vector -> np.array([1,2,3])

2 -> D: matrix -> np.array([ [1,2], [3,4] ])

3 -> D+:tensor ->np.arary([ [ [1,2] , [3,4] ] ])

Broadcasting , reshaping , dot products , slicing many functions rely on knowing the number of ndim of an array

```python
# 0 -Dimensional 
np.array('Tech')
```

```python
# 1 -Dimensional 
np.array([1,2,3])
```

```python
np.array(
     [ 
     
     [1,2], 
     [3,4] 

 ]
         )
```

```python
# 1- D Array
arr= np.array([0,1,2,3,4,5],dtype='int8')
print(arr)
print(arr.ndim) # number of Dimensional
print(arr.shape) # number of rows , number of Columns
print(arr.size) # number of rows * number of Columns
print(arr.dtype) # Data types
```

```python
# 1 - D array
arr= np.array([0,1,2,3,4,5],dtype='int8')
print(arr)
print(arr.ndim) # number of Dimensional
print(arr.shape) # number of rows , number of Columns
print(arr.size) # number of rows * number of Columns
print(arr.dtype) # Data types
```

```python
# 2 - D Array
arr = np.array([
                [1,2,3,4,5],
         [2,3,4,5,8]
],dtype='int8')

print(arr)
print(arr.ndim) # number of Dimensional
print(arr.shape) # number of rows , number of Columns
print(arr.size) # number of rows * number of Columns
print(arr.dtype) # Data types
```
