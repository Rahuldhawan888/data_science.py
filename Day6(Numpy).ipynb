# *Numpy dimensional*

- Dimensions determine how
- data is organized( eg.. )

0 -> D: Scaler -> np.array(5)

1 -> D: Vector -> np.array([1,2,3])

2 -> D: matrix -> np.array([ [1,2], [3,4] ])

3 -> D+:tensor ->np.arary([ [ [1,2] , [3,4] ] ])

Broadcasting , reshaping , dot products , slicing many functions rely on knowing the number of ndim of an array

```python
# 0 -Dimensional 
np.array('Tech')
```

```python
# 1 -Dimensional 
np.array([1,2,3])
```

```python
np.array(
     [ 
     
     [1,2], 
     [3,4] 

 ]
         )
```

```python
# 1- D Array
arr= np.array([0,1,2,3,4,5],dtype='int8')
print(arr)
print(arr.ndim) # number of Dimensional
print(arr.shape) # number of rows , number of Columns
print(arr.size) # number of rows * number of Columns
print(arr.dtype) # Data types
```

```python
# 1 - D array
arr= np.array([0,1,2,3,4,5],dtype='int8')
print(arr)
print(arr.ndim) # number of Dimensional
print(arr.shape) # number of rows , number of Columns
print(arr.size) # number of rows * number of Columns
print(arr.dtype) # Data types
```

```python
# 2 - D Array
arr = np.array([
                [1,2,3,4,5],
         [2,3,4,5,8]
],dtype='int8')

print(arr)
print(arr.ndim) # number of Dimensional
print(arr.shape) # number of rows , number of Columns
print(arr.size) # number of rows * number of Columns
print(arr.dtype) # Data types
```

# *Numpy flatten()*
* function is used to convert a multi-dimensional array into a 1D array.
*  It returns a copy of the original array collapsed into one dimension.

```python
arr = np.array([
    [12,  6, 83, 38, 43],
       [ 6, 63, 78, 51, 25],
       [67, 64, 71, 76, 43]
       ])
print(arr.shape)
print(arr.size)
```

```python
arr.flatten() # 2D to 1D
```

# *Array Operations*:-

```python
a = np.array([5,4,3])
b = np.array([1,4,3])

print(a+b) # 1D array addition
print(a-b) # 1D array subtraction
print(a*b) # 1D array multiplication
print(a/b) # 1D array division
print(a**b) # 1D array power
print(a%2) # 1D array modulus
```

```python
np.add(a,b) , np.subtract(a,b) , np.multiply(a,b) , np.divide(a,b) , np.power(a,b) , np.mod(a,2) , np.divmod(a,b)
```

```python
np.pow(a,2)
```

```python
# axis => 0 (row wise) ,axis => 1 (column wise)
np.sum([a,b],axis=0) , np.sum([a,b],axis=1)
```
