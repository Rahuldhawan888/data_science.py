# *While loop*
1. Initialized a variable
2. Check the condition =>
3. print the value
4. Increment / Decrement
5. (Small to big number or big to small number)


```python
a = 1 # starting point
while a <= 10:
    print("i'm hungry")
    a += 1
else:
     print("i'm not hungry")
```

```python
i = 0  #
while i <=10:  # Small to big  , 0 <=10 , 1 <=10 , 2<=10 , 3 <=10 , 4 <=10 , 5<=10, ..... 11 <=10(False)
    # print("I'm Hungry? with ",i ,"Time")
    print(f"I'm Hungry? with {i+1} Time")
    i +=1 # Increment by 1 ( Small to big )

else:
  print("I'm here with False Condition!!!")
```

```python
# Calculate sum of first n numbers :- 
total = 0
n = int(input())
while n >= 0:
    total += n
    n -= 1 # decrement
    print(f"sum of first number is :- {total}")
```

# *For loop*
* A for loop in Python is used to iterate over a sequence (such as a **list, tuple, string, or range , set and dictionary **), executing a block of code once for each item in the sequence. This is known as a definite loop because the number of iterations is determined by the length of the iterable.
