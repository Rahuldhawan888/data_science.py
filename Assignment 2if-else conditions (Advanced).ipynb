1. Hospital Emergency Room Triage

A hospital uses the following triage system for patients:
If the patient’s temperature is above 102°F or they report severe pain, classify them as "High
Priority.
"

If the temperature is between 99°F and 102°F or they report moderate pain, classify them as
"Medium Priority.
"

Otherwise, classify them as "Low Priority.
"

Take the patient’s temperature and pain level ("none,
" "moderate,
" or "severe") as input, and
display their priority.

```python
temperature = float(input("enter patient's temperature:°"))
pain = input("enter pain level (none, moderate, severe): ").lower().strip()

if temperature > 102 or pain == "severe":
    print("high priority")
elif 99 <= temperature <= 102 or pain == "moderate":
    print("medium priority")
else:
    print("low priority")
```

2. Shopping Cart Free Delivery

A store provides free delivery under the following conditions:

If the total cart value exceeds ₹2000.

If the total cart value is between ₹1000 and ₹2000 but the user has a
membership. Otherwise, the delivery fee is ₹100.

Take the cart value and membership status ("yes" or "no") as input, and determine if the
delivery is free or chargeable.

```python

cart_value = float(input("enter total cart value : "))
membership = input("do you have a membership? (yes/no): ").strip().lower()

if cart_value > 2000:
    print("free delivery")
elif 1000 <= cart_value <= 2000 and membership == "yes":
    print("free delivery")
else:
    print("delivery Fee: ₹100")
```

3. Library Fine System

A library charges a fine for overdue books as follows:

Up to 5 days overdue: ₹2 per day.

6–10 days overdue: ₹5 per day.

More than 10 days overdue: ₹10 per day.

Take the number of overdue days as input and calculate the fine.

```python
days = int(input("enter the number of overdue days: "))

if days <= 5:
    fine = days * 2
elif days <= 10:
    fine = days * 5
else:
    fine = days * 10

print("total fine:", fine)

```

4. Airport Luggage Check

Airlines allow the following luggage limits:

Economy Class: Up to 20 kg. Excess luggage is charged ₹200 per kg.

Business Class: Up to 30 kg. Excess luggage is charged ₹150 per kg.

First Class: Up to 40 kg. Excess luggage is charged ₹100 per kg.

Take the class type ("economy,
" "business," or "first") and the luggage weight as input, and
calculate the excess fee if applicable.

```python
class_type = input("enter class (economy,business,first): ").strip().lower()
weight = float(input("enter luggage weight (kg): "))

if class_type == "economy":
    if weight > 20:
        fee = (weight - 20) * 200
        print("excess Fee: ", fee)
    else:
        print("No excess fee.")

elif class_type == "business":
    if weight > 30:
        fee = (weight - 30) * 150
        print("excess Fee: ", fee)
    else:
        print("no excess fee.")

elif class_type == "first":
    if weight > 40:
        fee = (weight - 40) * 100
        print("excess Fee: ", fee)
    else:
        print("no excess fee.")

else:
    print("invalid class.")
```

5. Hotel Room Pricing

A hotel offers discounts based on the type of customer and room type:

Regular customers get 10% off on all rooms.

Members get 20% off on deluxe rooms and 15% off on standard rooms.

VIP customers get 30% off on all rooms.

Take customer type ("regular,
" "member,
" or "VIP") and room type ("standard" or "deluxe") as
input, and calculate the final price after applying the discount.

```python
customer = input("enter customer type:(regular , member , vip)").strip().lower()
room = input("enter room type:(standard , deluxe)").strip().lower()
price = float(input("enter room price: "))

if customer == "regular":
    discount = 10
elif customer == "member":
    if room == "deluxe":
        discount = 20
    else:
        discount = 15
elif customer == "vip":
    discount = 30

final_price = price - (price * discount / 100)
print("final price:", final_price)
```

6. Energy Consumption Billing

An electricity company charges based on energy consumption as follows:

For consumption up to 100 units, the rate is ₹5 per unit.

For consumption between 101 and 300 units, the rate is ₹10 per unit for additional
units. For consumption above 300 units, the rate is ₹15 per unit for additional units.

```python
units = int(input("enter total consumption units: "))

if units <= 100:
    bill = units * 5
elif units <= 300:
    bill = (100 * 5) + (units - 100) * 10
else:
    bill = (100 * 5) + (200 * 10) + (units - 300) * 15

print("total Bill:", bill)
```

7. Ride Fare Calculation

A ride-sharing app charges based on the time of day and distance:

Daytime (6 AM to 10 PM): ₹15 per km.

Nighttime (10 PM to 6 AM): ₹20 per km.

If the distance is more than 20 km, apply a 10% discount on the total fare. Take
the time ("day" or "night") and distance as input, and calculate the total fare

```python
time = input("enter time (day,night): ").strip().lower()
distance = float(input("enter distance in km: "))

if time == "day":
    fare = distance * 15
elif time == "night":
    fare = distance * 20
else:
    fare = 0
    print("invalid time")

if time == "day" or time == "night":
    if distance > 20:
        fare = fare * 0.9

    print("total fare: ", fare)
```

9. Parking Lot Charges

A parking lot charges based on the type of vehicle:

Cars: ₹50 for the first hour, ₹20 for every additional hour.

Bikes: ₹20 for the first hour, ₹10 for every additional hour.

Take the type of vehicle ("car" or "bike") and the parking duration as input, and calculate the
parking fee.

```python
vehicle = input("enter vehicle type (car,bike): ").strip().lower()
hours = int(input("enter parking duration in hours: "))

if vehicle == "car":
    if hours <= 1:
        fee = 50
    else:
        fee = 50 + (hours - 1) * 20
    print("parking Fee: ", fee)

elif vehicle == "bike":
    if hours <= 1:
        fee = 20
    else:
        fee = 20 + (hours - 1) * 10
    print("parking Fee: ", fee)

else:
    print("invalid vehicle type")
```

10. Real Estate Loan Eligibility

A bank offers real estate loans based on the following criteria:

If the applicant earns more than ₹1,00,000 per month and has no existing loans, they are
eligible for a 90% loan-to-value (L TV) ratio.

If the applicant earns between ₹50,000 and ₹1,00,000, or has existing loans, they are eligible
for a 70% L TV ratio.

Otherwise, they are eligible for a 50% L TV ratio.

Take the monthly income and loan status ("yes" or "no") as input, and display the eligible L TV
ratio.

```python
income = float(input("enter monthly income: "))
loan = input("do you have existing loans? (yes/no): ").strip().lower()

if income > 100000 and loan == "no":
    print("eligible LTV ratio: 90%")
elif (income >= 50000 and income <= 100000) or loan == "yes":
    print("eligible LTV ratio: 70%")
else:
    print("eligible LTV ratio: 50%")
```
