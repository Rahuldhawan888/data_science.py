# *IQR*:-


```python
min_val = df['total_price'].min
max_val = df['total_price'].max

min_val , max_val
```

```python
q1 = df[['total_price']].quantile(0.25) # 1st quartile
q2 = df[['total_price']].quantile(0.50) # 2st quartile
q3 = df[['total_price']].quantile(0.75) # 3st quartile

q1,q2,q3
```

```python
# IQR => inter quartile range 
# IQR = q3 - q1

# MIN range = q1 = 1.5 * IQR
# MAx range = q3 = 1.5 * IQR

iqr = q3 - q1 
print(iqr)

```

```python
min_range = q1 - 1.5 * iqr 
max_range = q3 - 1.5 * iqr 

print(min_range)
print(max_range)
```

# *Handle the null , missing , Duplicated*

```python
# Creating a DataFrame with various types of missing values
data = {
    'Age': [25,25, np.nan, 30, 35,np.nan], # NaN for numeric data
    'Name': ['A','A', '', 'A', None,"B"], # None for object/string data
    'Date': ['2023-01-01', '2023-01-01',"", '2023-01-03', None,'04-01-2023'] # NaT for dates
}

df = pd.DataFrame(data)
print(df)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
for i in df.columns:
    if df[i].dtype=='number':
        df[i] = df[i].replace("",np.nan) # missing
        df[i] = df[i].replace(" ",np.nan) # white space value
    elif df[i].dtype=='str':
        df[i] = df[i].replace("",None) # missing
        df[i] = df[i].replace(" " ,None) # white space value
```

```python
df['Date'] = pd.to_datetime(df['Date'],format='mixed')
```

```python
df.head()
```

```python
new_df = df.copy()
```

```python
new_df.isnull().sum()
```

```python
new_df.shape  , new_df.shape[0]  , len(new_df)
```

```python
new_df.isnull().sum() / new_df.shape[0]  * 100
```

# *Fillna / Dropna*

- Mean / Median for Numerical columns
- Mode for Category columns
- Forward / Backword fillna for the Date/time

- Email / commment / feedback => Default / Other value


```python
# use dropna
new_df.dropna()
```

```python
new_df.dropna(inplace=True) # Permanently on the original dataset 
```

```python
# use fillna 
new_df.head()
```

```python
# using Fillna 
new_df.head()
```

```python
new_df.info()
```

```python
for i in new_df.columns:
    if new_df[i].dtype=='float':
        new_df[i] = new_df[i].fillna(new_df[i].median())

    elif new_df[i].dtype=='str':
        new_df[i] = new_df[i].fillna(new_df[i].mode()[0])

    elif new_df[i].dtype=='datetime64[us]':
        new_df[i] = new_df[i].ffill()
        new_df[i] = new_df[i].bfill() 
```
