# *series :-*
* series is a one dimensional array which can hold any data type. 
* It is similar to a column in a table. It can be created using the pd.Series() function.


```python
# series 
arr = np.random.randint(1,100,10)
print(arr)
```

```python
s = pd.Series(arr,name='marks')
s
```

```python
s.mean(),s.median(),s.var(),s.std()
```

```python
s[0]
```

```python
# Series attributes
print(f"\nValues: {s.values}")
print(f"Index: {s.index}")
print(f"Data type: {s.dtype}")
print(f"Size: {s.size}")
print(f"Shape: {s.shape}")
print(f"Dimension: {s.ndim}")
print(f"Is unique: {s.is_unique}")
print(f"Is null: {s.isnull()}")
print(f"Is null values: {s.isnull().values}")
print(f"Is null any: {s.isnull().values.any()}")
print(f"Is null sum: {s.isnull().values.sum()}")
```

```python
sales= pd.Series([np.nan,3400,1200,6500,4500,3400,np.nan,None,""])
sales
```

```python
# Series attributes
print(f"\nValues: {sales.values}")
print(f"Index: {sales.index}")
print(f"Data type: {sales.dtype}")
print(f"Size: {sales.size}")
print(f"Shape: {sales.shape}")
print(f"Dimension: {sales.ndim}")
print(f"Is unique: {sales.is_unique}")
print(f"Is null: {sales.isnull()}")
print(f"Is null values: {sales.isnull().values}")
print(f"Is null any: {sales.isnull().values.any()}")
print(f"Is null sum: {sales.isnull().values.sum()}")
```
