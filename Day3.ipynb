# *list methods :-*

• *append()*: Adds an element to the end of the list.

• *extend()*: Appends elements from an iterable (like another list) to the end of the current list.
 
• *insert()*: Inserts an element at a specified index.

• *remove()*: Removes the first occurrence of a specific value.

• *pop()*: Removes and returns the element at a specified index (defaults to the last item).

• *clear()*: Removes all elements from the list, leaving it empty.

`Ordering and Sorting`

• *sort()*: Sorts the list items in place (ascending by default).

• *reverse()*: Reverses the order of the items in place.

`Information and Utility`

• *index()*: Returns the index of the first occurrence of a specific value.
• *count()*: Returns the number of times a specific value appears in the list.
• *copy()*: Returns a shallow copy of the list.

# *Copy*
*Deep copy vs Shallow Copy* 

# *Shallow copy*
* Both are dependent to eachother & Refer to the same object (means same memory location )

```python
li = [74,44,33,55,22,57]
new_li = li  # Shallow copy 

print(new_li) 
print(li) 

print(li is new_li)
print(li == new_li)
```

```python
new_li[3] = 74 # Replace value at index 3
```

```python
print(new_li)
print(li)
```

# *Deep Copy*:-
* Both are Independent to eachother & Not Refer to the same object (means both have not memory location )

```python
li = [74,44,33,55,22,57]
new_li = li.copy() # deep copy  

print(new_li) 
print(li) 

print(li is new_li)
print(li == new_li)
```

```python
li[0] = 44
print(new_li)
print(li)
```

# *Dictionary Methods*

* It is un-order and mapping data types.
* It have store key-value (pairs)
* It keys always will be unique
* It is also mutable ( key => immutable , values => mutable )
* Dictionary also supported different-different data types.

1. keys() — Get all keys
2. values() — Get all values
3. items() — Get key-value pairs
4. get() — Get a value
5. update() — Add or update items
6. pop() — Remove a specified key
7. popitem() — Remove the last inserted item    
8. clear() — Remove all items
9. copy() — Copy a dictionary
10. setdefault() — Get or insert a key

```python
student = {
    "name": ["Rahul","mahi"],
    "age": [20,23],
    "course": ["BA","BCA"]
}
```

```python
print(student)
```

# *Keys*

```python
print(student.keys()) # get all keys
```

# *Values*


```python
print(student.values()) # get items only 
```

# *Items*

```python
print(student.items()) # get value with pairs 
```

# *Get*

```python
print(student.get("name")) # get value you need 
```

# *update*

```python
student.update({"age":21}) # edit or update something 
print(student)
```

# *pop*

```python
student.pop("age") # Remove a specified key
print(student) 
```

# *popitem*

```python
student.popitem() # Remove the last inserted item
print(student)
```

# *copy*

```python
new_student = student.copy() # Copy a dictionary (deep copy)
print(new_student)
```

# *setdefault*

```python
student.setdefault("city","delhi") # add or insert a key 
print(student)
```

# *clear*

```python
student.clear() # Remove all items
print(student)
```
