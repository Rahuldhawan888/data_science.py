# *Conditional Statement*
( if, else , elif ) :-
1. `if` statement
Executes code when the condition is true.
2. `if-else` statement
Chooses between two options.
3. `if-elif-else` statement
Checks multiple conditions.


```python
sales = [45,76,24,75,24,86]

val = int(input("enter your sales\n"))

val in sales 
```

```python
if val in sales:
    print("hellow")
```

```python
if val in sales: 
    print(f"your last transaction found in my history")
else:
    print(f"your last transaction doesn's exit in my history")
```

```python
# check a bank balance is even , odd 
# positive , neutral , negative 
balance = float(input())
print(f"my bank balance is :- {balance}")
```

```python
if balance > 0 :
    print(f"this is my even number {balance}")
else:
    print(f"this is mt odd number {balance}")
```

```python
if balance > 0:
    print(f"this is my positive number {balance}")
elif balance > 0:
    print(f"this is my neutral number {balance}")
else:
    print(f"this is my negative number {balance}")
```
