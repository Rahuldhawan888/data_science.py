# *While loop*
1. Initialized a variable
2. Check the condition =>
3. print the value
4. Increment / Decrement
5. (Small to big number or big to small number)


```python
a = 1 # starting point
while a <= 10:
    print("i'm hungry")
    a += 1
else:
     print("i'm not hungry")
```

```python
i = 0  #
while i <=10:  # Small to big  , 0 <=10 , 1 <=10 , 2<=10 , 3 <=10 , 4 <=10 , 5<=10, ..... 11 <=10(False)
    # print("I'm Hungry? with ",i ,"Time")
    print(f"I'm Hungry? with {i+1} Time")
    i +=1 # Increment by 1 ( Small to big )

else:
  print("I'm here with False Condition!!!")
```

```python
a = 0 # Initialized
while a<=10: # Check conditions
    a +=1 # a = a + 1 =>  Increment by 1
    print(f"Index:- {a}")



# a = 0 , 0 <=10 Iteration:- 1
# a = 1 , 1 <=10 Iteration:- 2
# a = 2 , 2 <=10 Iteration:- 3
# a = 3 , 3 <=10 Iteration:- 4
# a = 4 , 4 <=10 Iteration:- 5
# a = 5 , 5 <=10 Iteration:- 6
# a = 6 , 6 <=10 Iteration:- 7
# a = 7 , 7 <=10 Iteration:- 8
# a = 8 , 8 <=10 Iteration:- 9
# a = 9  , 9 <=10 Iteration:- 10
# a = 10 , 10 <=10 Iteration:- 11
# a = 11 , 11 <=10 => False ( Automatical break the while block. )
```

```python
i =  10
while i >=0: # Big to Small   , 10 >=0 , 9 >=0 , 8 >=0 , 7 >=0 ,........ -1 >=0(False)
  print('I"m Hungry? with ',i,'Time')
  i -=1 # Decrement by 1 ( Big to Small )

else:
  # print('Condition is False')
   print("I'm here with False Condition!!!")
```

```python
# Calculate sum of first n numbers :- 
total = 0
n = int(input())
while n >= 0:
    total += n

    n -= 1 # decrement
    print(f"sum of first number is :- {total}")
```

```python
fact = 1
val = int(input())

while val >= 1:
    fact *= val
    val -= 1
    print(f"factorial result1 :- {fact}")
```

# *For loop*
* A for loop in Python is used to iterate over a sequence (such as a **list, tuple, string, or range , set and dictionary **), executing a block of code once for each item in the sequence. This is known as a definite loop because the number of iterations is determined by the length of the iterable.

```python
for val in range(5):
    print(val)
```

```python
for val in range(1,10):
    print(val)
```

```python
for val in range(1,10,2):
    print(val)
```

```python
# Negative
for VAL in range(10,-1,-1):
    print(VAL)
```

```python
li = [1200,1250,1500,900]

for i in li:
    print(i)
```

```python
for i in range(len(li)):
    print(f"index :- {i} and value {li[i]}")
```

```python
logs = ["error","info","error","waring","info","error"]
new_logs = set(logs)
print(new_logs)
for i in new_logs:
    print(logs.count(i))
```

```python
logs = ["error","info","error","waring","info","error"]

error = 0
info = 0
waring = 0

i = 0
for i in logs:
    if i =="error":
     error += 1
    elif i =="waring":
     waring +=1
    else:
      info += 1

print("Error",error)
print("Waring",waring)
print("info",info)
```

```python
sales = [1200, 2500, 1800, 3200, 900]

for amount in sales:
    if amount > 2000:
        print(amount, "greater than 2000")
    else :
        print(amount, "less than 2000")
```

```python
sales = [1200, 1500, 900, 2200, 1800]

for amount in sales:
    if amount > 2000:
        print(amount,"greater than 2000")
    elif amount < 1000:
        print(amount,"lowest than 1000")
    else :
        print(amount,"equals 1500")
        
    print("Total sales: ₹", sum(sales))
print("Average sales: ₹", sum(sales) / len(sales))
```
