# *Conditional Statement*
( if, else , elif ) :-
1. `if` statement
Executes code when the condition is true.
2. `if-else` statement
Chooses between two options.
3. `if-elif-else` statement
Checks multiple conditions.


```python
sales = [45,76,24,75,24,86]

val = int(input("enter your sales"))

val in sales 
```

```python
if val in sales:
    print("hellow")
```

```python
if val in sales: 
    print(f"your last transaction found in my history")
else:
    print(f"your last transaction doesn's exit in my history")
```

```python
# check a bank balance is even , odd 
# positive , neutral , negative 
balance = float(input("enter you balance:"))
if balance % 2 ==  0:
    print(f"this is my even number {balance}")
else:
    print(f"this is my odd number {balance}")

```

```python
if balance > 0:
    print(f"this is my positive number {balance}")
elif balance > 0:
    print(f"this is my neutral number {balance}")
else:
    print(f"this is my negative number {balance}")
```

1. Age-based condition checker ( Check people are eligible for vote or note )

5. Find the final price of each customers, 
- if customers selling price more than 5000 then calculate 5% discount for each selling price .

```python
#1
age = int(input("Enter your age: "))

if age >= 18:
    print(f"Eligible for vote: {age}")
else:
    print(f"Not eligible for vote: {age}")
```

```python
#2
age = int(input("Enter your age: "))
salary = float(input("Enter your monthly salary: "))

if age >= 25 and salary >= 25000:
    print(f"Eligible for loan {age,salary")
else:
    print(f"Not eligible for loan")
```

```python
#3 
age = int(input("Enter your age: "))

if age >= 18 and age <= 40:
    print(f"Age is between 18 and 40")
else:
    print(f"Age must be below 18 or above 40 ")
```

```python
#4 
income = float(input("enter your income"))
               
if income <= 100000 :
               print(f"income must be 1L {income}")
else:
               print(f"income below 1L {income}")
               
```

```python
#5 
price = float(input("enter your price of item:"))

if price > 50000 :
    discount_price = price + (price * 0.05)
    print(f"your total amount pay is :{discount_price}")
else:
    print(f"our total amount pay is : {price}")

```

```python
#6 
a = int(input("enter your num:"))
b = int(input("enter your num:"))           
c = int(input("enter your num:"))

if a > b and b > c:
          print(f"{a} is greater than {b} and {c}")
elif b > a and a > c:
          print(f"{b} is greater than {a} and {c}")
else:
          print(f"{c} is greater than {a} and {b}")
```
